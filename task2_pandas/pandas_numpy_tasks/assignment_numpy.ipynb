**1.** Дан случайный массив, поменять знак у элементов, значения которых между 3 и 8

```python
arr = np.random.randint(0, 12, 20)
print("Original array: ", arr)

condition = (arr > 3) & (arr < 8)
indexes = np.where(condition)[0]
values = arr[condition]

print("Indexes meeting condition: ", indexes)
print("Values meeting condition: ", values)

arr[condition] = -values

print("Modified array: ", arr)
```

**3.** Построить прямое произведение массивов (все комбинации с каждым элементом). На вход подается двумерный массив

**4.** Даны 2 массива A (8x3) и B (2x2). Найти строки в A, которые содержат элементы из каждой строки в B, независимо от порядка элементов в B

```python
B = np.random.randint(0, 10, [2,2])
A = np.random.randint(0, 10, [8,3])
print(A)
print(B)

result = []
for i, row_a in enumerate(A):
    set_a = set(row_a)
    for row_b in B:
        if set(row_b).issubset(set_a):
            result.append(i)
            break

print("A indexes:", result)
```

**5.** Дана 10x3 матрица, найти строки из неравных значений (например строка [2,2,3] остается, строка [3,3,3] удаляется)

```python

A = np.random.randint(0, 2, [10, 3])
print(A)

unique = np.array([len(np.unique(row)) for row in A])

res_ind = np.where(unique>1)[0]
res_rows = A[unique>1]

print("\nIndexes of diff rows:", res_ind)

print("\nResult:\n", res_rows)
```

**6.** Дан двумерный массив. Удалить те строки, которые повторяются

```python
size = np.random.randint(2,10)
A = np.random.randint(0, 2, [size,size])
print(A)

unique = np.array([len(np.unique(row)) for row in A])

res_ind = np.where(unique>1)[0]
res_rows = A[unique>1]

print("\nIndexes of diff rows:", res_ind)

print("\nResult:\n", res_rows)
```

* __Задача 4__: Реализовать кодирование длин серий (Run-length encoding). Для некоторого вектора x необходимо вернуть кортеж из двух векторов одинаковой длины. Первый содержит числа, а второй - сколько раз их нужно повторить.  
 Например, для x = np.array([2, 2, 2, 3, 3, 3, 5]) ответ (np.array([2, 3, 5]), np.array([3, 3, 1])).

```python
size=randint(2,15)
B = [ randint(0,7) for _ in range(size)]
print(B)
values = []
lengths = []
current_val = B[0]  
count = 1           
for val in B[1:]:   
    if val == current_val:
        count += 1  
    else:
        values.append(current_val)  
        lengths.append(count)      
        current_val = val          
        count = 1                  
values.append(current_val)  
lengths.append(count)       
print("Res: ", values, lengths)
```
